```python
#Statistical Analysis and Visualization of DJIA dataset

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv(r"C:\Users\DELL\Downloads\djia.csv")
df.head
```

```python
mean_price = df['Price'].mean()
median_price = df['Price'].median()
std_price = df['Price'].std()
min_price = df['Price'].min()
max_price = df['Price'].max()

print("Mean DJIA: ", mean_price)
print("Median DJIA: ", median_price)
print("Standard Deviation of DJIA: ", std_price)
print("Min DJIA: ", min_price)
print("Max DJIA: ", max_price)
```

```python
df.boxplot(column='Price')

plt.title('Box Plot of DJIA Prices')
plt.xlabel('DJIA Price')
plt.show()
```

```python
plt.figure()

sns.violinplot(x='Price', data=df)

plt.title('Violin Plot of DJIA Prices')
plt.xlabel('DJIA Price')
plt.ylabel('Density')

plt.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df.sort_values('Date', inplace=True)

df.plot(x='Date', y='Price', kind='line')

plt.title('DJIA Over Time')
plt.xlabel('Date')
plt.ylabel('DJIA Price')
plt.show()
```
